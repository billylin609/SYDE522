# **Binary Classification using Gradient Descent**

Let's create 6 2D points manually and spread them across the 4 quadrants to force the data to be split. This is a simple example. When we learn SVM we learn how to deal with complex classification examples as the gradient descent on its own is not enough to increase the complexity.

```python
import numpy as np
import matplotlib.pyplot as plt

# Preprocessing Input data
X_1 = np.array((1, -3, 1, 2, 3, 0))
X_2 = np.array((3, 0, -1, -2, -1, 6))
Y = np.array((-1, -1, 1, 1, 1, -1))

data = np.vstack([X_1, X_2])
plt.scatter(X_1, X_2, c=Y)
plt.show()
```

Building the model

```python
# Weights to be learned
w1 = -1 # try (-1, -1) to make it bad
w2 = -1

alpha = 0.01 # learning Rate - try 0.0001 with 500 epochs to make it slow
epochs = 500 # The number of iterations - try 10 and 50 to make it bad then 500

n = float(len(X_1))
```

Training the model using the Gradient Descent algorithm

```python
for i in range(epochs):
    score = w1*X_1 + w2*X_2 # you can use it to monitor training
    Y_pred = np.sign(score) # you can use it to monitor training
    training_set = np.vstack((X_1, X_2, Y)).T
    d_w1 = (1/n) * sum(-x1 * y if 1 - (w1*x1 + w2*x2) * y > 0 else 0 for x1, x2, y in training_set) # gradient wrt w1
    d_w2 = (1/n) * sum(-x2 * y if 1 - (w1*x1 + w2*x2) * y > 0 else 0 for x1, x2, y in training_set) # gradient wrt w2
    w1 = w1 - alpha * d_w1  # update w1
    w2 = w2 - alpha * d_w2  # update w2
    print(f'epoch {i}: w = {w1, w2}, gradientF = {d_w1, d_w2}')
```

Plotting the final model

```python
print('Final trained weights:', w1, w2)

score = w1*X_1 + w2*X_2
Y_pred = np.sign(score)

plt.scatter(X_1, X_2, c=Y)

x_line = np.linspace(min(X_1), max(X_1), 100)
plt.plot(x_line, -(w1/w2) * x_line, color='red')

plt.show()
```
